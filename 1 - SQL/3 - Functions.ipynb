# Functions

No SQL Server, uma _function_ (ou função) é um objeto de banco de dados que permite encapsular um conjunto de instruções SQL para realizar uma tarefa específica. As funções podem ser usadas para retornar valores ou conjuntos de resultados e podem ser chamadas em consultas, instruções `SELECT`, `WHERE`, entre outras.

```sql
-- Craindo uma função para formatação de data usando o DATENAME

CREATE FUNCTION FnDataCompleta(@data as DATE)
RETURNS VARCHAR(MAX)
AS 
BEGIN
    RETURN DATENAME(DW, @Data) + ', '+
            DATENAME(D, @Data) + ' de ' +
            DATENAME(M, @Data) + ' de ' +
            DATENAME(Y, @Data) 

END

-- > Alterar function
--ALTER FUNCTION...

-- > Apagar function
-- DROP FUNCTION ...
```

```sql
CREATE OR ALTER FUNCTION FnDataCompleta(@data as DATE)
RETURNS VARCHAR(MAX)
AS 
BEGIN
    RETURN DATENAME(DW, @Data) + ', '+
            DATENAME(D, @Data) + ' de ' +
            DATENAME(M, @Data) + ' de ' +
            DATENAME(Y, @Data) + ' - ' +
            CASE
            WHEN MONTH(@Data) <= 6 THEN '(1º Semestre)'
            ELSE '(1º Semestre)' 
            END 
END
```

```sql
--> Utilizando a function
SELECT 
nome_cliente,
data_de_nascimento,
[dbo].[FnDataCompleta](data_de_nascimento) as DataCompleta
FROM
dCliente

```
