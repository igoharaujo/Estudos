# Pivo Table

"PIVOT table" ou tabela pivot é uma técnica usada para transformar dados de uma estrutura de linhas em uma estrutura de colunas, facilitando a análise dos dados, especialmente para relatórios e dashboards.  

Basicamente, o `PIVOT` permite sumarizar e reorganizar os dados, tornando valores de uma coluna específica em cabeçalhos de colunas e aplicando uma função agregada, como `SUM`, `AVG`, `COUNT`, entre outros, nos valores de outras colunas.

```sql
use ContosoRetailDW
```

```sql
-- Exemplo:

SELECT
    departementname,
    YEAR(hiredate) Ano,
    COUNT(Employeekey)  total_funcionarios
FROM
    dimemployee
GROUP BY departmentname, year(hiredate)
```
