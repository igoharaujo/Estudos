# OPERAÇÕES CRUD NO MongoDB ATLAS VIA PYTHON

### Conectando o Python ao MongoDB Atlas

```python
# Importar biblioteca
from pymongo import MongoClient

# Pegar URI de conexão no MongoDB Atlas.
# Alterar <db_password> pela senha criada no capítulo anterior, tópico "COMO CRIAR UM CLUSTER GRATUITO PARA TESTE"
MONGODB_URI = "mongodb+srv://hashtagsql:<db_password>@cluster0.fxce4.mongodb.net/?retryWrites=true&w=majority&appName=Cluster0"

# Conectar
conexao = MongoClient(MONGODB_URI)
print("Conexão bem sucedida!")
```

#### insert_one(): inserindo um documento

```python
# pega o nome do BD (se não existir, o MongoDB cria o BD na hora de inserir)
banco = conexao.hashtag

# pega o nome da coleção (se não existir, o MongoDB cria a coleção na hora de inserir)
colecao = banco.cursos

# variável que recebe o novo documento
novo_curso = {
    "curso": "SQL Impressionador",
    "ano_lancamento": 2021,
    "categoria": "Programação",
    "cursos_recomendados": ["Power BI Impressionador", "Python Impressionador"]
}

# insere o novo documento na coleção
resultado = colecao.insert_one(novo_curso)

# pega o _id do novo documento inserido acima
id_documento = resultado.inserted_id

#imprime na tela o _id do novo documento inserido acima
print(f"Documento inserido com sucesso!\nNúmero do _id: {id_documento}")
```

#### insert_many(): inserindo mais de um documento

```python
# pega o nome do BD (se não existir, o MongoDB cria o BD na hora de inserir)
banco = conexao.hashtag

# pega o nome da coleção (se não existir, o MongoDB cria a coleção na hora de inserir)
colecao = banco.cursos

# variável que recebe os novos documentos
novos_cursos = [
    {
    "curso": "Power BI Impressionador",
    "ano_lancamento": 2016,
    "categoria": "Microsoft",
    "cursos_recomendados": ["Excel Impressionador", "SQL Impressionador"]
    },
    {
    "curso": "Python Impressionador",
    "ano_lancamento": 2020,
    "categoria": "Programação"
    }
]

# insere os novos documentos na coleção
resultado = colecao.insert_many(novos_cursos)

# pega os _id dos novos documentos inseridos acima
ids_documentos = resultado.inserted_ids

#imprime na tela a quantidade e os _id dos novos documentos inseridos acima
print(f"Documentos inseridos com sucesso!\nNúmeros do _ids: {ids_documentos}")
```

#### find_one(): retorna um documento

```python
# NO ATLAS, MOSTRAR QUE TEMOS 3 DOCUMENTOS NA COLEÇÃO

# pega o nome do BD
banco = conexao.hashtag

# pega o nome da coleção
colecao = banco.cursos

# variável que recebe o filtro do documento a ser consultado
encontrar_documento = {"ano_lancamento": {"$gte": 2020}}

# procura o documento filtrado na coleção
resultado = colecao.find_one(encontrar_documento)

# imprime na tela o documento encontrado
print(resultado)
```

#### find(): retorna mais de um documento

```python
# pega o nome do BD
banco = conexao.hashtag

# pega o nome da coleção
colecao = banco.cursos

# variável que recebe o filtro dos documentos a serem consultados
encontrar_documentos = {"ano_lancamento": {"$gte": 2020}}

# procura o documento filtrado na coleção
resultado = colecao.find(encontrar_documentos)

# imprime na tela os documentos encontrados e a quantidade de documentos encontrados:
for documento in resultado:
    print(documento)
```

#### update_one(): atualiza um documento

```python
# pega o nome do BD
banco = conexao.hashtag

# pega o nome da coleção
colecao = banco.cursos

# variável que recebe o filtro do documento a ser atualizado
atualizar_documento = {"curso": "Python Impressionador"}

# variável que recebe a atualização a ser feita no documento filtrado
# (operador $set acrescenta campo cursos_recomendados / ou atualiza seu valor, se já existir)
novo_campo = {"$set": {"cursos_recomendados": ["SQL Impressionador", "JavaScript Impressionador"]}}

# imprime o documento original (antes da atualização)
print(colecao.find_one(atualizar_documento))

# atualiza o documento e mostra na tela uma mensagem de confirmação
resultado = colecao.update_one(atualizar_documento, novo_campo)
print("\nDocumento atualizado: " + str(resultado.modified_count))

# imprime o documento atualizado
print(colecao.find_one(atualizar_documento))
```

#### update_many(): atualiza mais de um documento

```python
# pega o nome do BD
banco = conexao.hashtag

# pega o nome da coleção
colecao = banco.cursos

# variável que recebe o filtro dos documentos a serem atualizados
atualizar_documentos = {"categoria": "Programação"}

# variável que recebe a atualização a ser feita nos documentos filtrados 
# (atualizar o campo cursos_recomendados para "IA Impressionador" e "JavaScript Impressionador" para os documentos da categoria "Programação"
# (operador $set atualiza valor do campo cursos_recomendados / ou cria novo campo, se não existir)
atualiza_campo = {"$set": {"cursos_recomendados": ["Inteligência Artificial Impressionador", "JavaScript Impressionador"]}}

# atualiza os documentos
resultado = colecao.update_many(atualizar_documentos, atualiza_campo)

# imprime quantos documentos foram encontrados (matched_count) e quantos foram atualizados (modified_count)
print("Documentos encontrados: " + str(resultado.matched_count))
print("Documentos atualizados: " + str(resultado.modified_count))

# imprime na tela os documentos atualizados (iterando o cursor com um for)
for documento in colecao.find(atualizar_documentos):
    print(documento)
```

#### delete_one(): exclui um documento

```python
# pega o nome do BD
banco = conexao.hashtag

# pega o nome da coleção
colecao = banco.cursos

# variável que recebe o filtro do documento a ser excluído
excluir_documento = {"categoria": "Microsoft"}

# imprime o documento a ser excluído
print(colecao.find_one(excluir_documento))

# exclui o documento 
resultado = colecao.delete_one(excluir_documento)

# imprime a quantidade de documentos excluídos
print("\nDocumento excluído: " + str(resultado.deleted_count))
```

#### delete_many(): exclui mais de um documento

```python
# pega o nome do BD
banco = conexao.hashtag

# pega o nome da coleção
colecao = banco.cursos

# variável que recebe o filtro dos documentos a serem excluídos
excluir_documentos = {"categoria": "Programação"}

# imprime o documento a ser excluído
for documento in colecao.find(excluir_documentos):
    print(documento)

# exclui os documentos
resultado = colecao.delete_many(excluir_documentos)

# imprime a quantidade de documentos excluídos
print("\nDocumentos excluídos: " + str(resultado.deleted_count))
```
